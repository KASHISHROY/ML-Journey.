```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import warnings
warnings.filterwarnings("ignore")
```

```python
df=pd.read_csv('Travel.csv')
```

```python
df['Gender']=df['Gender'].replace('Fe male','Female')
df['MaritalStatus']=df['MaritalStatus'].replace('Single','Unmarried')
```

```python
feature_with_na=[feature for feature in df.columns if df[feature].isnull().sum()>=1]
print(feature_with_na)
```

```python
##misising values handled
df.Age.fillna(df.Age.median(),inplace=True)
```

```python
df.TypeofContact.fillna(df.TypeofContact.mode(),inplace=True)
df.DurationOfPitch.fillna(df.DurationOfPitch.median(),inplace=True)
df.NumberOfFollowups.fillna(df.NumberOfFollowups.mode(),inplace=True)
df.PreferredPropertyStar.fillna(df.PreferredPropertyStar.mode(),inplace=True)
df.NumberOfChildrenVisiting.fillna(df.NumberOfChildrenVisiting.mode(),inplace=True)
df.MonthlyIncome.fillna(df.MonthlyIncome.median(),inplace=True)
```

```python
df.NumberOfTrips.fillna(df.NumberOfTrips.median(),inplace=True)
```

```python
##feature engineering
df['TotalVisiting']=df['NumberOfPersonVisiting']+df['NumberOfChildrenVisiting']
```
