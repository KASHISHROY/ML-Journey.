```python
##SUPPORT VECTOR REGRESSION IMPLEMENTATION
##DATESET->TIPS DATASET WHICH IS PRESENT IN SEABORN LIBRARY
```

```python
import seaborn as sns
df=sns.load_dataset('tips')
```

```python
df.head()    #total bill is the dependent feature
```

```python
df.info()
```

```python
df['sex'].value_counts()   #sex,smoker,time->binary category->label encoding,day ->3 categories one hot encoding
```

```python
df.columns
```

```python

##independent and dependent features
X=df[['tip', 'sex', 'smoker', 'day', 'time', 'size']]
y=df['total_bill']
```

```python
#train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=0)
```

```python
X_train.head()
```

```python
##Feature Encoding(Label encoding and onehot encoding)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
#3 encoders for 3 binary catregories
le1=LabelEncoder()
le2=LabelEncoder()
le3=LabelEncoder()
```

```python
import warnings
warnings.filterwarnings('ignore')
X_train['sex']=le1.fit_transform(X_train['sex'])
X_train['smoker']=le2.fit_transform(X_train['smoker'])
X_train['time']=le3.fit_transform(X_train['time'])
```

```python
X_train.head()
```

```python
##same thing on test dataset
X_test['sex']=le1.fit_transform(X_test['sex'])
X_test['smoker']=le2.fit_transform(X_test['smoker'])
X_test['time']=le3.fit_transform(X_test['time'])
```

```python
X_test.head()
```

```python
##one hot encoding---using Column transformer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct=ColumnTransformer(transformers=[('onehot',OneHotEncoder(drop='first'),[3])],remainder='passthrough')
```

```python

X_train=ct.fit_transform(X_train)
```

```python
X_test=ct.fit_transform(X_test)
```

```python
##SVR--SUPPORT VECTOR REGRESSION
from sklearn.svm import SVR
svr=SVR()
```

```python
svr.fit(X_train,y_train)
```

```python
y_pred=svr.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
print(r2_score(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
```
