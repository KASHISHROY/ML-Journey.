```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
##load the dataset
from sklearn.datasets import load_breast_cancer
```

```python
cancer_dataset=load_breast_cancer()
```

```python
cancer_dataset.keys()
```

```python
print(cancer_dataset.DESCR)
```

```python
##data key has the entire data->create  a dataframe using that
df=pd.DataFrame(cancer_dataset['data'],columns=cancer_dataset['feature_names'])
```

```python
df.head()
```

```python
##Standardization->good step to do before applying feature extraction
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
scaler.fit(df)
```

```python
scaled_data=scaler.transform(df)
```

```python
scaled_data
```

```python
##Applying pca algorithm from all these feature we will extracted only 2 features
from sklearn.decomposition import PCA
pca=PCA(n_components=2) ##no of features you want to have 
```

```python
data_pca=pca.fit_transform(scaled_data)
```

```python
data_pca  ##now has only 2 features and max variance is getting captured
```

```python
pca.explained_variance_   ##2 variances will get
```

```python
##if we dont get components in pca then the no of features it has earlier that no of components will be present after words too and variance also will have how much feature we had
pca1=PCA()
```

```python
pca1.fit_transform(scaled_data)
```

```python
pca1.explained_variance_
```

```python
##VISUALIZATION
plt.figure(figsize=(8,6))
plt.scatter(data_pca[:,0],data_pca[:,1],c=cancer_dataset['target'],cmap='plasma')
plt.xlabel('FIRST PRINCIPAL COMPONET')
plt.ylabel('SECOND PRINCIPLE COMPONET')
```
