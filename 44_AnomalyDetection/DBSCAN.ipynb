```python
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
%matplotlib inline
from sklearn.datasets import make_circles
```

```python
X,y=make_circles(n_samples=750,factor=0.3,noise=0.1)

##we are getting y but we wont use the y,y==output feature,we will just use the x apply clusteribg and try to find out how many clusters formed and whethere it is same as the no of unique values of y present
```

```python
plt.scatter(X[:,0],X[:,1],c=y)

##how our real data points look like
```

```python
from sklearn.cluster import DBSCAN
```

```python
dbscan=DBSCAN(eps=0.10)
```

```python
dbscan.fit_predict(X)   #wherever value -1 those are the outliers
```

```python
plt.scatter(X[:,0],X[:,1],c=dbscan.labels_)
##dark purple one are the outliers->this is the clustering done by dbscan of our datapoints
```
